# Q7

```python
balance = 50000

while True:
    print("\n===== ATM MENU =====")
    print("1. Check Balance")
    print("2. Deposit")
    print("3. Withdraw")
    print("4. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        print(f"Current Balance: Rs. {balance}")

    elif choice == "2":
        amount = float(input("Enter deposit amount: Rs. "))

        if amount > 0:
            balance = balance + amount
            print("Deposit Successful")
            print(f"New Balance: Rs. {balance}")
        else:
            print("Invalid Amount")

    elif choice == "3":
        amount = float(input("Enter withdrawal amount: Rs. "))

        if amount <= 0:
            print("Invalid Amount")
        elif amount > balance:
            print("Insufficient Balance")
        else:
            balance = balance - amount
            print("Withdrawal Successful")
            print(f"Remaining Balance: Rs. {balance}")

    elif choice == "4":
        print("Thank you for using ATM.")
        break

    else:
        print("Invalid Choice")
```

```python
num_students = int(input("Enter number of students: "))

passed = 0
failed = 0
class_total_marks = 0
class_total_percentage = 0

highest_percentage = None
lowest_percentage = None

for i in range(num_students):
    print(f"\n===== Student {i + 1} =====")

    name = input("Enter student name: ")
    python_marks = float(input("Enter Python marks: "))
    math_marks = float(input("Enter Mathematics marks: "))
    english_marks = float(input("Enter English marks: "))

    total_marks = python_marks + math_marks + english_marks
    percentage = (total_marks / 300) * 100

    print(f"Student: {name}")
    print(f"Total Marks: {total_marks}/300")
    print(f"Percentage: {percentage:.2f}%")

    if (percentage >= 50 and python_marks >= 40
            and math_marks >= 40 and english_marks >= 40):

        result = "Pass"
        passed = passed + 1

        if percentage >= 90:
            grade = "A+"
        elif percentage >= 80:
            grade = "A"
        elif percentage >= 70:
            grade = "B"
        elif percentage >= 60:
            grade = "C"
        else:
            grade = "D"

    else:
        result = "Fail"
        grade = "No Grade"
        failed = failed + 1

    print(f"Result: {result}")
    print(f"Grade: {grade}")

    class_total_marks = class_total_marks + total_marks
    class_total_percentage = class_total_percentage + percentage

    if highest_percentage is None or percentage > highest_percentage:
        highest_percentage = percentage

    if lowest_percentage is None or percentage < lowest_percentage:
        lowest_percentage = percentage

print("\n===== CLASS REPORT =====")
print(f"Total Students: {num_students}")
print(f"Passed: {passed}")
print(f"Failed: {failed}")
print(f"Class Total Marks: {class_total_marks:.2f}")

if num_students > 0:
    class_average = class_total_percentage / num_students

    print(f"Highest Percentage: {highest_percentage:.2f}%")
    print(f"Lowest Percentage: {lowest_percentage:.2f}%")
    print(f"Class Average: {class_average:.2f}%")
else:
    print("No students entered.")
```
