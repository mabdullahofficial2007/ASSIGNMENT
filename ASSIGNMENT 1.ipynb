# 2. DATA TYPES

```python
My_String = "HELLO FAST" #string data type
My_Int = 18              #integar data type
My_Float = 18.5          #float data type
My_Bool = True           #boolean data type

print("STRING :",My_String)
print("INTEGER :",My_Int)
print("FLOAT :",My_Float)
print("BOOLEAN :",My_Bool)

print("<----------DATA_TYPE----------->")
print(type(My_String))
print(type(My_Int))
print(type(My_Float))
print(type(My_Bool))
```

# 3. ARITHMETIC OPERATER

```python
Num1 = 5
Num2 = 2

print("ADDITION :" ,Num1 + Num2)
print("SUBTRACTION :" ,Num1 - Num2)
print("MULTIPLICATION :" ,Num1 * Num2)
print("DIVISION :" ,Num1 / Num2)
print("MODULE :" ,Num1 % Num2)
print("POWER :" ,Num1 ** Num2)
```

# 6. RECTANGLE AREA

```python
LENGTH = 5
WIDTH = 10

AREA_RECTANGLE = LENGTH * WIDTH
PERIMETER_RECTANGLE = 2 * (LENGTH + WIDTH)

print("LENGTH :", LENGTH)
print("WIDTH :", WIDTH)
print("RECTANGLE AREA :", AREA_RECTANGLE)
print("RECTANGLE PERIMETER :", PERIMETER_RECTANGLE)
```

# 7. TIME CONVERTER

```python
TOTAL_MINUTE = 250

Hours = TOTAL_MINUTE // 60
Remaining_Minute = TOTAL_MINUTE % 60

print("TOTAL MINUTE :", TOTAL_MINUTE)
print("HOURS :", Hours)
print("REMAINING MINUTE :", Remaining_Minute)
```

# 8. STUDENT ELIGIBILITY

```python
MARKS = 90
ATTENDENCE = 65

IS_ELIGIBLE = (MARKS >= 50) and (ATTENDENCE >= 75)

print("MARKS :",MARKS)

print("ATTENDENCE :",ATTENDENCE)

print("ELIGIBLE FOR EXAM :",IS_ELIGIBLE)
```

# 9. LOGICAL OPERATOR

# 10. STUDENT REPORT

```python
STUDENT_NAME = "ABDULLAH"
AGE = 18
PHY_MARKS = 85
COMP_MARKS = 92
MATH_MARKS = 90
ENG_MARKS = 71
URDU_MARKS = 89
ATTENDENCE = 100

TOTAL_MARKS = PHY_MARKS + COMP_MARKS + MATH_MARKS + ENG_MARKS + URDU_MARKS
AVERAGE_MARKS = TOTAL_MARKS/5

IS_ELIGIBLE = (AVERAGE_MARKS >= 50) and (ATTENDENCE >= 75)

print("STUDENT_NAME :", STUDENT_NAME ,
      "\nAGE :", AGE, 
      "\nCOMP_MARKS :", COMP_MARKS ,
      "\nMATH_MARKS :", MATH_MARKS , 
      "\nENG_MARKS :", ENG_MARKS ,
      "\nURDU_MARKS :", URDU_MARKS ,
      "\nATTENDENCE :", ATTENDENCE )

print("\nTOTAL MARKS : ",TOTAL_MARKS)
print("AVERAGE MARKS :", AVERAGE_MARKS)
print("STUDENT ELIGIBLE FOR SCHOLARSHIP :", IS_ELIGIBLE)
```
