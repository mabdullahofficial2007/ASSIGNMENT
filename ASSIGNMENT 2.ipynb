```python
amount = float(input("Enter total purchase amount: "))

if amount < 5000:
    discount_rate = 0
elif amount < 10000:
    discount_rate = 0.10
elif amount < 20000:
    discount_rate = 0.15
else:
    discount_rate = 0.20

discount = amount * discount_rate
final_amount = amount - discount

print(f"Original Amount: Rs. {amount:.2f}")
print(f"Discount: Rs. {discount:.2f}")
print(f"Final Amount: Rs. {final_amount:.2f}")
```

# Q11

```python
name = input("Enter student name: ")
age = int(input("Enter age: "))
matric = float(input("Enter Matric percentage: "))
intermediate = float(input("Enter Intermediate percentage: "))
entry_test = float(input("Enter entry test score: "))
domicile = input("Do you have a valid domicile? (yes/no): ").lower()

print("\n===== ADMISSION REPORT =====")
print(f"Name: {name}")
print(f"Age: {age}")
print(f"Intermediate: {intermediate}%")
print(f"Entry Test: {entry_test}%")

# Basic eligibility
if (age >= 17 and matric >= 60 and intermediate >= 60
        and entry_test >= 50 and domicile == "yes"):

    # Merit category
    if entry_test >= 85 and intermediate >= 85:
        merit = "Excellent Merit"
        scholarship = "50% Scholarship"

    elif entry_test >= 70 and intermediate >= 70:
        merit = "Good Merit"
        scholarship = "25% Scholarship"

    elif entry_test >= 50 and intermediate >= 60:
        merit = "Normal Merit"
        scholarship = "No Scholarship"

    else:
        merit = "Not Eligible"
        scholarship = "No Scholarship"

    if merit != "Not Eligible":
        print("Status: Eligible")
        print(f"Merit: {merit}")
        print(f"Scholarship: {scholarship}")
    else:
        print("Status: Not Eligible")
        print(f"Merit: {merit}")
        print(f"Scholarship: {scholarship}")

else:
    print("Status: Not Eligible")

    if age < 17:
        print("Reason: Age must be at least 17.")

    if matric < 60:
        print("Reason: Matric percentage must be at least 60%.")

    if intermediate < 60:
        print("Reason: Intermediate percentage must be at least 60%.")

    if entry_test < 50:
        print("Reason: Entry test score must be at least 50%.")

    if domicile != "yes":
        print("Reason: Valid domicile is required.")
```

# Q12

```python
name = input("Enter employee name: ")
experience = float(input("Enter years of experience: "))
salary = float(input("Enter monthly salary: Rs. "))
performance = float(input("Enter performance score out of 100: "))
projects = int(input("Enter number of projects completed: "))
late_days = int(input("Enter number of late days: "))

print("\n===== EMPLOYEE REPORT =====")
print(f"Employee: {name}")
print(f"Experience: {experience} years")
print(f"Monthly Salary: Rs. {salary:.2f}")
print(f"Performance: {performance}")
print(f"Projects: {projects}")
print(f"Late Days: {late_days}")

# Promotion eligibility
if (experience >= 2 and performance >= 70
        and projects >= 3 and late_days < 10):

    print("Promotion Status: Eligible")

    # Promotion level
    if performance >= 90 and experience >= 5:
        promotion = "Senior Promotion"

    elif performance >= 80 and experience >= 3:
        promotion = "Mid-Level Promotion"

    else:
        promotion = "Junior Promotion"

    # Bonus category
    if performance >= 90:
        bonus = "20%"
    elif performance >= 80:
        bonus = "15%"
    elif performance >= 70:
        bonus = "10%"
    else:
        bonus = "No bonus"

    print(f"Promotion Level: {promotion}")
    print(f"Bonus: {bonus}")

else:
    print("Promotion Status: Not Eligible")

    if experience < 2:
        print("Reason: Experience must be at least 2 years.")

    if performance < 70:
        print("Reason: Performance score must be at least 70.")

    if projects < 3:
        print("Reason: At least 3 projects must be completed.")

    if late_days >= 10:
        print("Reason: Late days must be less than 10.")
```
